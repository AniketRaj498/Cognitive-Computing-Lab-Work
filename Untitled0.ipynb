```python
import pandas as pd
import numpy as np
import random

random.seed(21)
np.random.seed(21)

n1 = 10
loan_data = {
    'Tid': list(range(1, n1 + 1)),
    'Refund': [random.choice(['Yes', 'No']) for _ in range(n1)],
    'Marital_Status': [random.choice(['Single', 'Married', 'Divorced']) for _ in range(n1)],
    'Taxable_Income': [random.randint(50, 250) * 1000 for _ in range(n1)],
    'Cheat': [random.choice(['Yes', 'No']) for _ in range(n1)]
}
loan_df = pd.DataFrame(loan_data)

print("Loan dataframe:")
print(loan_df)


print("\nrows 0, 3, 6, 9 only:")
print(loan_df.iloc[[0, 3, 6, 9]])


print("\nrows 2 to 6:")
print(loan_df.iloc[2:7])


print("\nrows 3-7, cols 1-3:")
print(loan_df.iloc[3:8, 1:4])


print("\nall rows, cols 1-3:")
print(loan_df.iloc[:, 1:4])



names = ['Aman', 'Riya', 'Karan', 'Priya', 'Vivek', 'Neha', 'Sahil', 'Anjali']
departments = ['HR', 'IT', 'Marketing', 'Sales', 'Finance']
n2 = 8

emp_data = {
    'Employee_ID': [201 + i for i in range(n2)],
    'Name': names,
    'Department': [random.choice(departments) for _ in range(n2)],
    'Age': [random.randint(22, 45) for _ in range(n2)],
    'Salary': [random.randint(30000, 90000) for _ in range(n2)],
    'Years_of_Experience': [random.randint(1, 15) for _ in range(n2)],
    'Rating': [round(random.uniform(3.0, 5.0), 1) for _ in range(n2)]
}
emp_df = pd.DataFrame(emp_data)

print("\n\nEmployee dataframe:")
print(emp_df)

print("\nshape ->", emp_df.shape)

print("\ninfo:")
emp_df.info()

print("\ndescribe:")
print(emp_df.describe())

print("\nfirst 3 rows:")
print(emp_df.head(3))
print("last 2 rows:")
print(emp_df.tail(2))


print("\navg salary:", emp_df['Salary'].mean())
print("total salary:", emp_df['Salary'].sum())
print("youngest age:", emp_df['Age'].min())
print("best rating:", emp_df['Rating'].max())


print("\nsorted by salary (high to low):")
print(emp_df.sort_values(by='Salary', ascending=False))



def cat_rating(r):
    if r >= 4.5:
        return 'Excellent'
    elif r >= 4.0:
        return 'Good'
    else:
        return 'Average'



emp_df['Performance_Category'] = emp_df['Rating'].apply(cat_rating)
print("\nafter adding Performance_Category:")
print(emp_df)

print("\nnull values per column:")
print(emp_df.isnull().sum())

emp_df = emp_df.rename(columns={'Employee_ID': 'ID'})
print("\ncolumns after rename:", emp_df.columns.tolist())


print("\nexperience > 5 years:")
print(emp_df[emp_df['Years_of_Experience'] > 5])

print("\nonly IT department:")
print(emp_df[emp_df['Department'] == 'IT'])

emp_df['Tax'] = emp_df['Salary'] * 0.10
print("\nwith Tax column added:")
print(emp_df)

print("\ncount per department:")
print(emp_df.groupby('Department').size())

emp_df.to_csv('employees_output.csv', index=False)
reloaded = pd.read_csv('employees_output.csv')
print("\nread back from the csv file:")
print(reloaded.head())


arr = np.random.randint(10, 100, size=12).reshape(3, 4)
print("\n\nnumpy array:")
print(arr)

print("\nmean:", np.mean(arr))
print("median:", np.median(arr))
print("max:", np.max(arr))
print("min:", np.min(arr))
print("unique values:", np.unique(arr))

reshaped = arr.reshape(4, 3)
print("\nreshaped to 4x3:")
print(reshaped)

resized = np.resize(arr, (2, 3))
print("\nresized to 2x3 (some values got dropped):")
print(resized)

lin_arr = np.linspace(10, 100, 10)
print("\nlinspace array:")
print(lin_arr)
print("shape:", lin_arr.shape, "| dims:", lin_arr.ndim, "| total elements:", lin_arr.size)
```
