```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("Adding 2:", arr + 2)
print("Multiplying by 3:", arr * 3)
print("Dividing by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

unique_x, counts_x = np.unique(x, return_counts=True)
max_count_x = np.max(counts_x)
frequent_x = unique_x[counts_x == max_count_x]

print("Most frequent value in x:", frequent_x)
print("Indices in x:", np.where(np.isin(x, frequent_x))[0])

unique_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
frequent_y = unique_y[counts_y == max_count_y]

print("Most frequent value in y:", frequent_y)
print("Indices in y:", np.where(np.isin(y, frequent_y))[0])
```

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
ucs420_your_name = np.linspace(10, 100, 25)

print("Array:", ucs420_your_name)
print("Dimensions:", ucs420_your_name.ndim)
print("Shape:", ucs420_your_name.shape)
print("Total elements:", ucs420_your_name.size)
print("Data type:", ucs420_your_name.dtype)
print("Total bytes:", ucs420_your_name.nbytes)

transposed_ucs420_your_name = ucs420_your_name.reshape(25, 1)
print("Transpose using reshape():")
print(transposed_ucs420_your_name)

transposed_using_T = ucs420_your_name.T
print("Transpose using T:")
print(transposed_using_T)

print("Yes, T can be used, but for a 1-D array T does not change its shape.")
print("To get a column vector, reshape(25, 1) is required.")
```

```python
ucs420_your_name = np.array([[10, 20, 30, 40],
                             [50, 60, 70, 80],
                             [90, 15, 20, 35]])

print("Original Array:")
print(ucs420_your_name)

print("Mean:", np.mean(ucs420_your_name))
print("Median:", np.median(ucs420_your_name))
print("Maximum:", np.max(ucs420_your_name))
print("Minimum:", np.min(ucs420_your_name))
print("Unique elements:", np.unique(ucs420_your_name))

reshaped_ucs420_your_name = ucs420_your_name.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_your_name)

resized_ucs420_your_name = np.resize(reshaped_ucs420_your_name, (2, 3))

print("Resized Array (2 x 3):")
print(resized_ucs420_your_name)
```
