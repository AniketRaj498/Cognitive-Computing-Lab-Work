```python
import pandas as pd
import numpy as np

# Q.6
emp_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}
emp_df = pd.DataFrame(emp_data)
emp_df.to_csv('employees.csv', index=False)

employees = pd.read_csv('employees.csv')

shape_of_df = employees.shape
print("Shape:", shape_of_df)

print("Summary:")
employees.info()

desc_stats = employees.describe()
print(desc_stats)

print(employees.head())
print(employees.tail(3))

avg_salary = employees['Salary'].mean()
total_bonus = employees['Bonus'].sum()
youngest_age = employees['Age'].min()
highest_rating = employees['Rating'].max()

sorted_emp = employees.sort_values(by='Salary', ascending=False)

conditions = [
    (employees['Rating'] >= 4.5),
    (employees['Rating'] >= 4.0) & (employees['Rating'] < 4.5),
    (employees['Rating'] < 4.0)
]
choices = ['Excellent', 'Good', 'Average']
employees['Performance_Category'] = np.select(conditions, choices, default='')

missing_values = employees.isnull().sum()
print(missing_values)

employees = employees.rename(columns={'Employee_ID': 'ID'})

experienced_employees = employees[employees['Years_of_Experience'] > 5]
it_employees = employees[employees['Department'] == 'IT']

employees['Tax'] = employees['Salary'] * 0.10

employees.to_csv('modified_employees.csv', index=False)
```
